## Scientific notation

Engineering calculations often contain values that are extremely large or
extremely small.

Examples include:

- Radio frequencies measured in millions or billions of hertz
- Signal periods measured in microseconds or nanoseconds
- Satellite distances measured in thousands or millions of metres
- Very small received powers

Scientific notation represents a number as:

$$
a \times 10^n
$$

where:

- $a$ is the coefficient
- $n$ is the exponent
- $10^n$ determines the scale of the value

Python uses the letter `e` to represent the power of ten.

Examples:

```python
1e3    # 1 × 10³
1e6    # 1 × 10⁶
1e9    # 1 × 10⁹

1e-3   # 1 × 10⁻³
1e-6   # 1 × 10⁻⁶
1e-9   # 1 × 10⁻⁹

````python
# Scientific notation in Python 

# Positive exponents represent large values. 

one_kilohertz_hz = 1e3
one_megahertz_hz = 1e6
one_gigahertz_hz = 1e9

# Negative exponents represent small values.

one_millisecond_s = 1e-3
one_microsecond_s = 1e-6
one_nanosecond_s = 1e-9

print("1 kHz =", one_kilohertz_hz, "Hz")
print("1 MHz =", one_megahertz_hz, "Hz")
print("1 Ghz =", one_gigahertz_hz, "Hz")

print("1 ms =", one_millisecond_s, "s")
print("1 microsecond =", one_microsecond_s, "s")
print("1 ns =", one_nanosecond_s, "s")

print(type(one_gigahertz_hz))
print(type(one_nanosecond_s))
````

## Operator Precedence in Engineering Calculations

Operator precedence determines the order in which Python evaluates mathematical operations.

The main precedence rules used in this course are:

1. Parentheses: `()`
2. Exponents: `**`
3. Multiplication and division: `*`, `/`
4. Addition and subtraction: `+`, `-`

Operations with the same precedence are generally evaluated from left to right.

For engineering calculations, it is good practice to use parentheses even when they are not strictly required. Parentheses make equations easier to read and reduce the risk of calculation errors.

For example, the signal period is

\[
T = \frac{1}{f}
\]

In Python, this should be written as:

```python
T = 1 / f

````python
a = 10
b = 5
c = 3

result_without_parentheses = a + b / c
result_with_parentheses = (a + b) / c

print("Without parentheses:", result_without_parentheses)
print("With parentheses:", result_with_parentheses)
````

## Parentheses in Engineering Equations

Parentheses preserve the mathematical structure of an engineering equation when it is translated into Python.

Consider the equation:

$$
x = \frac{a+b}{c+d}
$$

The numerator and denominator are complete expressions. Therefore, both must be enclosed in parentheses:

```python
x = (a + b) / (c + d)
```

Writing the expression without parentheses changes its mathematical meaning:

```python
x = a + b / c + d
```

Python interprets this expression as:

$$
x = a + \frac{b}{c} + d
$$

This happens because division has a higher operator precedence than addition.

Another common engineering expression is:

$$
y = \frac{a-b}{2c}
$$

A clear Python implementation is:

```python
y = (a - b) / (2 * c)
```

Multiplication must be written explicitly in Python. Therefore, the mathematical term (2c) becomes:

2 * c




````python
# Demonstrating parentheses in engineering equations 

# Example input values 
a = 12 
b = 4
c = 3
d = 1 

# Correct translation of the engineering equation: 
# x = (a + b) / (c + d)

correct_result = (a + b) / (c + d)

# Incorrect translation caused by missing parentheses: 
# Python evaluates division before addition 

incorrect_result = a + b / c + d

# Display the results 
print("Correct result:", correct_result)
print("Result without required parentheses:", incorrect_result)
````

## Formatted Output with f-Strings

Engineering calculations should produce output that is clear, readable, and includes the correct physical units.

Python **f-strings** allow variables and calculated values to be inserted directly into text.

An f-string begins with the letter `f` before the opening quotation mark:

```python
f"Text {variable}"
```

The expression inside the braces `{}` is evaluated by Python and inserted into the output.

For example:

```python
frequency_hz = 2_400_000_000
print(f"Frequency: {frequency_hz} Hz")
```

This produces:

```text
Frequency: 2400000000 Hz
```

### Controlling Decimal Places

Engineering results often contain many decimal places. An f-string can control how many decimal places are displayed.

The format:

```python
{value:.2f}
```

displays the value using two digits after the decimal point.

For example:

```python
distance_km = 35786.123456
print(f"Distance: {distance_km:.2f} km")
```

produces:

```text
Distance: 35786.12 km
```

Other common formats include:

* `{value:.1f}` — one decimal place
* `{value:.2f}` — two decimal places
* `{value:.3f}` — three decimal places
* `{value:.6f}` — six decimal places
* `{value:.3e}` — scientific notation with three decimal places

### Thousands Separators

Large numerical values can be made easier to read by adding commas:

```python
{value:,}
```

For example:

```python
print(f"Frequency: {frequency_hz:,} Hz")
```

produces:

```text
Frequency: 2,400,000,000 Hz
```

A value can use both thousands separators and decimal-place formatting:

```python
{value:,.2f}
```

### Engineering Output Practice

A useful engineering output should normally contain:

1. A descriptive quantity name
2. The numerical result
3. An appropriate number of decimal places
4. The physical unit

For example:

```python
print(f"Satellite altitude: {altitude_km:,.2f} km")
```

The formatting changes only how a value is displayed. It does not change the value stored in the variable.


````python
# Formatted engineering output with f-strings 

# Example engineering quantities 
frequency_hz = 2_400_000_000
satellite_altitutde_km = 35_786.123456
signal_period_s = 0.0000000004167

# Basic f-string output 
print(f"Frequency: {frequency_hz} Hz")

# Add a thousands separator 
print(f"Formatted frequency: {frequency_hz:,} Hz") # comma separates the thousands as formatting

# Control the number of decimal places
print(f"Satellite altitude: {satellite_altitutde_km:.2f} km") # floating number formatting

# Combine a thousands separator with decimal-place formatting 
print(f"Formatted satellite altitude: {satellite_altitutde_km:,.2f} km") # thousand and decimal place formatting

# Display a very small value using scientific notation 
print(f"Signal period: {signal_period_s:.3e} s")
````

## SI Prefixes in RF and Space Engineering

Engineering quantities can be extremely large or extremely small. SI prefixes provide a compact way to represent powers of ten.

For example:

$$
1\ \text{GHz} = 10^9\ \text{Hz}
$$

and:

$$
1\ \text{ns} = 10^{-9}\ \text{s}
$$

Common SI prefixes used in RF, radar, satellite communications, and satellite navigation include:

| Prefix | Symbol | Power of Ten |                              Example |
| ------ | -----: | -----------: | -----------------------------------: |
| giga   |      G |       $10^9$ |    $1\ \text{GHz} = 10^9\ \text{Hz}$ |
| mega   |      M |       $10^6$ |    $1\ \text{MHz} = 10^6\ \text{Hz}$ |
| kilo   |      k |       $10^3$ |    $1\ \text{kHz} = 10^3\ \text{Hz}$ |
| milli  |      m |    $10^{-3}$ |   $1\ \text{ms} = 10^{-3}\ \text{s}$ |
| micro  |      µ |    $10^{-6}$ | $1\ \mu\text{s} = 10^{-6}\ \text{s}$ |
| nano   |      n |    $10^{-9}$ |   $1\ \text{ns} = 10^{-9}\ \text{s}$ |
| pico   |      p |   $10^{-12}$ |  $1\ \text{ps} = 10^{-12}\ \text{s}$ |

### Converting to the Base SI Unit

To convert a prefixed quantity to its base SI unit, multiply it by the corresponding power of ten.

For a frequency of $2.4\ \text{GHz}$:

$$
f = 2.4 \times 10^9\ \text{Hz}
$$

For a time interval of $5\ \mu\text{s}$:

$$
t = 5 \times 10^{-6}\ \text{s}
$$

In Python, powers of ten can be written using scientific notation:

```python
frequency_hz = 2.4e9
time_s = 5e-6
```

The notation `2.4e9` means:

$$
2.4 \times 10^9
$$

The notation `5e-6` means:

$$
5 \times 10^{-6}
$$

### Case Sensitivity

SI prefix symbols are case-sensitive.

For example:

* `M` represents mega, or $10^6$
* `m` represents milli, or $10^{-3}$

Therefore, MHz and mHz represent very different frequencies:

$$
1\ \text{MHz} = 10^6\ \text{Hz}
$$

$$
1\ \text{mHz} = 10^{-3}\ \text{Hz}
$$

Using the correct uppercase or lowercase symbol is essential in engineering calculations.


````python
# SI prefixes used in RF and space engineering 

# Prefix scale factors 
kilo = 1e3
mega = 1e6
giga = 1e9 

milli = 1e-3
micro = 1e-6
nano = 1e-9
pico = 1e-12 

# Example engineering quantities 
frequency_ghz = 2.3 
bandwidth_mhz = 20 
sampling_rate_khz = 125 
pulse_duration_us = 5 
delay_ns = 250 

# Convert each quantity to is base SI unit 
frequency_hz = frequency_ghz * giga 
bandwidth_hz = bandwidth_mhz * mega
sampling_rate_hz = sampling_rate_khz * kilo
pulse_duration_s = pulse_duration_us * micro 
delay_s = delay_ns * nano 

# Display the converted values 
print(f"Frequency: {frequency_ghz} GHz = {frequency_hz:,.0f} Hz")
print(f"Bandwidth: {bandwidth_mhz} MHz = {bandwidth_hz:,.0f} Hz")
print(f"Sampling rate: {sampling_rate_khz} kHz = {sampling_rate_hz:,.0f} Hz")
print(f"Pulse duration: {pulse_duration_us} µs = {pulse_duration_s:.3e} s")
print(f"Propagation delay: {delay_ns} ns = {delay_s:.3e} s")
````

## Frequency Conversions

Frequency describes how many complete cycles occur per second.

The SI unit of frequency is the hertz:

$$
1\ \text{Hz} = 1\ \text{cycle per second}
$$

RF, radar, satellite communication, and satellite navigation systems commonly use kilohertz, megahertz, and gigahertz because their operating frequencies are much larger than one hertz.

### Common Frequency Units

| Unit      | Symbol | Equivalent in Hertz |
| --------- | -----: | ------------------: |
| hertz     |     Hz |      $1\ \text{Hz}$ |
| kilohertz |    kHz |   $10^3\ \text{Hz}$ |
| megahertz |    MHz |   $10^6\ \text{Hz}$ |
| gigahertz |    GHz |   $10^9\ \text{Hz}$ |

### Converting to Hertz

To convert a frequency to hertz, multiply by the corresponding SI scale factor.

For kilohertz:

$$
f_{\text{Hz}} = f_{\text{kHz}} \times 10^3
$$

For megahertz:

$$
f_{\text{Hz}} = f_{\text{MHz}} \times 10^6
$$

For gigahertz:

$$
f_{\text{Hz}} = f_{\text{GHz}} \times 10^9
$$

For example, a carrier frequency of $1.57542\ \text{GHz}$ is:

$$
f = 1.57542 \times 10^9\ \text{Hz}
$$

Therefore:

$$
f = 1{,}575{,}420{,}000\ \text{Hz}
$$

### Converting from Hertz

To convert from hertz to a larger frequency unit, divide by the corresponding scale factor.

For kilohertz:

$$
f_{\text{kHz}} = \frac{f_{\text{Hz}}}{10^3}
$$

For megahertz:

$$
f_{\text{MHz}} = \frac{f_{\text{Hz}}}{10^6}
$$

For gigahertz:

$$
f_{\text{GHz}} = \frac{f_{\text{Hz}}}{10^9}
$$

For example:

$$
2{,}400{,}000{,}000\ \text{Hz}
==============================

\frac{2{,}400{,}000{,}000}{10^9}\ \text{GHz}
$$

Therefore:

$$
2{,}400{,}000{,}000\ \text{Hz}
==============================

2.4\ \text{GHz}
$$

### Direct Conversion Between MHz and GHz

Because:

$$
1\ \text{GHz} = 1000\ \text{MHz}
$$

a value in gigahertz can be converted to megahertz by multiplying by 1000:

$$
f_{\text{MHz}} = f_{\text{GHz}} \times 1000
$$

A value in megahertz can be converted to gigahertz by dividing by 1000:

$$
f_{\text{GHz}} = \frac{f_{\text{MHz}}}{1000}
$$

### Engineering Practice

For calculations involving signal period or wavelength, frequency should normally be converted to hertz first.

This ensures that the units remain consistent with:

$$
T = \frac{1}{f}
$$

and:

$$
\lambda = \frac{c}{f}
$$

When frequency is expressed in hertz, the resulting period is in seconds and the resulting wavelength is in metres.


````python
# Frequency conversions for RF and space systems 

# SI frequency scale factors 
kilo = 1e3
mega = 1e6
giga = 1e9 

# Example frequency values 
frequency_khz = 125.0 
frequency_mhz = 1575.42
frequency_ghz = 10.0 

# Convert kHz, MHz and GHz to Hz 
frequency_from_khz_hz = frequency_khz * kilo 
frequency_from_mhz_hz = frequency_mhz * mega
frequency_from_giga_hz = frequency_ghz * giga 

# Display frequencies converted to hertz 
print(f"{frequency_khz} KHz = {frequency_from_khz_hz:,.0f} Hz")
print(f"{frequency_mhz} MHz = {frequency_from_mhz_hz:,.0f} Hz")
print(f"{frequency_ghz} GHz = {frequency_from_giga_hz:,.0f} Hz")

print()

# Convert a frequency in hertz to larger units 
frequency_hz = 2_400_000_000

frequency_khz_converted = frequency_hz / kilo 
frequency_mhz_converted = frequency_hz / mega 
frequency_ghz_converted = frequency_hz / giga 

# Display the converted frequency 
print(f"{frequency_hz:,} Hz = {frequency_khz_converted:,.0f} KHz")
print(f"{frequency_hz:,} Hz = {frequency_mhz_converted:,.0f} MHz")
print(f"{frequency_hz:,} Hz = {frequency_ghz_converted:.2f} GHz")
````
